# **SIMPLE INTEREST CALCULATOR**

````python
# Ask the user to enter the principal amount
principal = float(input("Enter the principal amount: "))

# Ask the user to enter the rate of interest
rate = float(input("Enter the rate of interest (%): "))

# Ask the user to enter the time period
time = float(input("Enter the time (in years): "))


# Formula for Simple Interest
# SI = (P * R * T) / 100

simple_interest = (principal * rate * time) / 100

# Calculate the total amount after interest
total_amount = principal + simple_interest


# Display the results

print("\n====== SIMPLE INTEREST RESULTS ======")

print(f"Principal Amount: {principal}")
print(f"Rate of Interest: {rate}%")
print(f"Time Period: {time} years")

print(f"\nSimple Interest: {simple_interest}")

print(f"Total Amount After Interest: {total_amount}")
````

### **Python Lists**

A list in Python is a built-in data structure used to store multiple items in a single variable.

Lists are:

- Ordered → items keep their position
- Mutable → items can be changed
- Dynamic → items can be added or removed
- Indexed → each item has a position starting from 0
- Flexible → can store different data types together

1. Creating Lists

````python
# Example 1: A List of Names

students = ["Brian", "Amina", "John", "Faith"]

print(students)
````

````python
# Example 2: A List of Numbers

numbers = [10, 20, 30, 40, 50]

print(numbers)
````

````python
# Example 3: Mixed Data Types

mixed_list = ["Rohan", 21, 69.75, True]

print(mixed_list)
````

2. Understanding List Indexing

Each item has a position called an index.

Item	Index

- Apple   0
- Banana	1
- Mango 	2
- Orange	3

````python
fruits = ["Apple", "Banana", "Mango", "Orange"]

print(fruits[0])
print(fruits[2])
````

3. Negative Indexing

Python allows counting from the end using negative numbers.

- Item	Negative Index
- Orange	-1
- Mango	-2
- Banana	-3

````python
fruits = ["Apple", "Banana", "Mango", "Orange"]

print(fruits[-1])
print(fruits[-2])
````

4. Slicing Lists

Slicing means extracting part of a list.

Syntax

list[start:end]

- start is included
- end is excluded

````python
numbers = [10, 20, 30, 40, 50, 60]

print(numbers[1:4])
````

More Slicing Examples

````python
numbers = [10, 20, 30, 40, 50]

print(numbers[:3])    # From beginning to index 2
print(numbers[2:])    # From index 2 to end
print(numbers[-3:])   # Last 3 items
````

5. Updating List Items

Lists are mutable, meaning values can be changed.

````python
subjects = ["Math", "Physics", "Chemistry"]

print("Before update:")
print(subjects)

subjects[1] = "Biology"

print("After update:")
print(subjects)
````

6. Adding Items to a List

a) append()

Adds one item to the end.

````python
students = ["Brian", "Amina"]

students.append("Kevin")

print(students)
````

b) insert()

Adds an item at a specific position.

````python
students = ["Brian", "Amina"]

students.insert(1, "Kevin")

print(students)
````

c) extend()

Adds multiple items.

````python
numbers = [1, 2, 3]

numbers.extend([4, 5, 6])

print(numbers)
````

7. Removing Items from a List

a) remove()

Removes a specific item.

````python
fruits = ["Apple", "Banana", "Mango"]

fruits.remove("Banana")

print(fruits)
````

b) pop()

Removes an item using its index.

````python
numbers = [10, 20, 30, 40]

removed_item = numbers.pop(2)

print("Removed:", removed_item)
print(numbers)
````

c) del

Deletes items completely.

````python
numbers = [1, 2, 3, 4]

del numbers[1]

print(numbers)
````

d) clear()

Removes everything from the list.

````python
numbers = [1, 2, 3]

numbers.clear()

print(numbers)
````

9. Useful List Methods

count()

Counts occurrences of an item.

````python
numbers = [1, 2, 2, 2, 3]

print(numbers.count(2))
````

index()

Finds the position of an item.

````python
fruits = ["Apple", "Banana", "Mango"]

print(fruits.index("Banana"))
````

reverse()

Reverses the list.

````python
numbers = [1, 2, 3, 4]

numbers.reverse()

print(numbers)
````

sort()

Sorts items in ascending order.

````python
numbers = [50, 10, 40, 20]

numbers.sort()

print(numbers)
````

Sorting in Descending Order

````python
numbers = [50, 10, 40, 20]

numbers.sort(reverse=True)

print(numbers)
````

10. Built-in Functions with Lists

max() and min()

````python
numbers = [5, 8, 2, 20]

print(max(numbers))
print(min(numbers))
````

sum()

Adds all numbers in a list.

````python
numbers = [10, 20, 30]

print(sum(numbers))
````
